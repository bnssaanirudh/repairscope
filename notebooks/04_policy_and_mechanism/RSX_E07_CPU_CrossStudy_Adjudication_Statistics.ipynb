# RSX-E07 — Cross-study adjudication, evidence audit and statistical readiness

**CPU only. No model generation.** Runs on the completed V1, G, NB29H, P1, E02 and E03 records. Produces an immutable evidence inventory, a human-review queue for every NB29H definite error, labeled hypotheses rather than optimistic reclassifications, and bootstrap summaries grouped at the problem level.

Does not modify historical labels or imply that already inspected held-out items are pristine confirmation cases.

```python
import json,base64,zlib,hashlib,collections,csv,io,math,statistics,zipfile,datetime,random
from pathlib import Path
from IPython.display import display
import pandas as pd
DATA=json.loads(zlib.decompress(base64.b64decode("")))
assert {k:len(DATA[k]) for k in ('V1','G','H','P1','E02','E03')}=={'V1':272,'G':32,'H':324,'P1':48,'E02':72,'E03':72}
ROOT=Path('/content/RSX_E07_Outputs');ROOT.mkdir(parents=True,exist_ok=True)
def dump(name,obj):
    p=ROOT/name;p.write_text(json.dumps(obj,indent=2,sort_keys=True,ensure_ascii=False,default=str));return p
print('Loaded historical records:',{k:len(DATA[k]) for k in ('V1','G','H','P1','E02','E03')})
```

```python
# Protocol integrity, uniqueness and visible potential selection leakage.
rows=[]
for group in ('V1','G','H','P1','E02','E03'):
    recs=DATA[group]; ids=[r.get('trial_id') for r in recs]
    if any(x is None for x in ids) or len(set(ids))!=len(ids):
        raise AssertionError(f'Missing or duplicate trial ID: {group}')
    counts=collections.Counter(r.get('problem_id') for r in recs)
    for r in recs:
        ev=r.get('evaluation',{})
        outcome=(ev.get('failure_type') or ev.get('outcome') or ('correct' if ev.get('success') else 'unknown'))
        rows.append({'study':group,'trial_id':r['trial_id'],'problem_id':r.get('problem_id'),
         'split':r.get('split', 'not_listed'),'kind':r.get('kind','unknown'),
         'outcome':outcome,'stop_reason':r.get('stop_reason'),
         'trial_count_for_problem':counts[r.get('problem_id')],
         'signature':r.get('protocol_hash',r.get('signature',''))})
frame=pd.DataFrame(rows);frame.to_csv(ROOT/'all_trial_inventory.csv',index=False)
report=frame.groupby(['study','outcome'],dropna=False).size().reset_index(name='n')
report.to_csv(ROOT/'outcomes_by_study.csv',index=False);display(report)
print('Potentially reused V1 heldout problem IDs in NB29H:',len(set(frame.query('study=="V1" and split=="heldout"').problem_id)&set(frame.query('study=="H"').problem_id)))
print('NB29H heldout-labeled records must be treated as exploratory if used to inform policy design.')
```

```python
# All NB29H definite failures: explicit audit flags, no silent correction of labels.
records=[r for r in DATA['H'] if r.get('evaluation',{}).get('eligible_semantic_failure')]
assert len(records)==23
review=[]
for r in sorted(records,key=lambda r:(r['problem_id'],r['mode'],r['budget'])):
    ev=r['evaluation']; text=str(r.get('output',''))
    flags=[]
    if ev.get('explicit_final_number') is None:flags.append('missing_numeric_final')
    if ev.get('wrong_expression'):flags.append('wrong_expression_reported')
    if r['kind']=='countdown':flags.append('manual_expression_validation')
    if r.get('stop_reason')=='length':flags.append('length_truncated')
    if r['split']=='heldout':flags.append('exploratory_heldout_exposed')
    if any(s in text[-30:] for s in ('\boxed{','FINAL_ANSWER:','ANSWER:')):flags.append('incomplete_terminal_marker')
    review.append({'trial_id':r['trial_id'],'problem_id':r['problem_id'],'split':r['split'],
       'kind':r['kind'],'mode':r['mode'],'budget':r['budget'],'original_failure_label':ev['failure_type'],
       'audit_flags':';'.join(flags),'proposed_manual_category':'REVIEW_REQUIRED',
       'observed_output':text})
pd.DataFrame(review).to_csv(ROOT/'H23_manual_adjudication.csv',index=False)
print('H candidate errors',len(review),'unique problems',len(set(r['problem_id'] for r in review)))
print('Ambiguity and truncation remain distinct from definite error.')
```

```python
# Recompute evidence summaries using independent problem clusters, not repeated seeds as IID cases.
from collections import defaultdict
H=DATA['H']
summary=[]
for mode in sorted({r['mode'] for r in H}):
 for budget in sorted({r['budget'] for r in H}):
    subset=[r for r in H if r['mode']==mode and r['budget']==budget]
    # One record per unique problem for this arm.
    byid={r['problem_id']:r for r in subset};assert len(subset)==len(byid)
    success=[int(r['evaluation'].get('success',False)) for r in byid.values()]
    n=len(success);k=sum(success)
    z=1.96;denom=1+z*z/n;cen=(k/n+z*z/(2*n))/denom
    half=z*math.sqrt((k/n)*(1-k/n)/n+z*z/(4*n*n))/denom
    summary.append({'mode':mode,'budget':budget,'unique_problems':n,'correct':k,'rate':k/n,'wilson_low':max(0,cen-half),'wilson_high':min(1,cen+half),
      'wrong':sum(r['evaluation']['failure_type']=='definite_wrong' for r in subset),
      'ambiguous':sum(r['evaluation']['failure_type']=='unanswered_or_ambiguous' for r in subset)})
pd.DataFrame(summary).to_csv(ROOT/'H_arms_wilson.csv',index=False)
# Paired bootstrap at problem level for concise vs structured, 96-token arms.
a={r['problem_id']:int(r['evaluation']['success']) for r in H if r['mode']=='concise' and r['budget']==96}
b={r['problem_id']:int(r['evaluation']['success']) for r in H if r['mode']=='structured' and r['budget']==96}
ids=sorted(set(a)&set(b));assert len(ids)==54
rr=random.Random(7367)
diffs=[]
for _ in range(4000):
 sample=rr.choices(ids,k=len(ids))
 diffs.append(sum(a[p]-b[p] for p in sample)/len(sample))
diffs.sort()
bootstrap={'contrast':'concise_minus_structured_96','independent_problem_count':len(ids),
 'observed_difference':sum(a[p]-b[p] for p in ids)/len(ids),
 'cluster_bootstrap_ci95':[diffs[100],diffs[3899]],'resamples':4000,
 'interpretation':'exploratory, selected cases, not heldout confirmation'}
dump('H_problem_bootstrap.json',bootstrap)
print(bootstrap)
```

```python
# P1 comparison grouped by unique problem (avoid treating identical-seed outputs as independent evidence).
p1=pd.DataFrame([{'problem_id':r['problem_id'],'action':r['action'],'seed':r['seed'],
 'success':int(r['evaluation']['success']),'loop_token_passes':r['loop_token_passes'],
 'wall_seconds':r['wall_seconds']} for r in DATA['P1']])
p1.groupby('action').agg(trials=('success','size'),unique_problems=('problem_id','nunique'),success_trials=('success','sum'),
 forward_passes=('loop_token_passes','sum')).reset_index().to_csv(ROOT/'P1_calibration_baselines.csv',index=False)
# Counterfactual sensitivity calculations, not power guarantees.
sens=[]
for n in [20,40,80,120,200]:
 for delta in [.05,.10,.20]:
    se=math.sqrt(.25/n) # illustration using worst-case Bernoulli variance
    sens.append({'independent_problems':n,'hypothetical_abs_gain':delta,'worst_case_approx_se':se,
     'gain_in_standard_errors':delta/se,'warning':'not a power analysis for paired clustered comparisons'})
pd.DataFrame(sens).to_csv(ROOT/'planning_sensitivity_illustrative.csv',index=False)
summary_json={'audit':'complete','input_study_counts':{k:len(DATA[k]) for k in ('V1','G','H','P1','E02','E03')},
 'H_wrong_trials':len(records),'H_distinct_wrong_problems':len({r['problem_id'] for r in records}),
 'P1_completed':len(DATA['P1']),'E02_completed':len(DATA['E02']),'E03_completed':len(DATA['E03']),
 'confirmation_ready':False,'note':'All H/E02/E03/P1 interventions are exploratory; no original label changed.'}
dump('AUDIT_SUMMARY.json',summary_json)
OUT=Path('/content/RSX_E07_CPU_Results.zip')
with zipfile.ZipFile(OUT,'w',zipfile.ZIP_DEFLATED) as zz:
 for x in sorted(ROOT.iterdir()):zz.write(x,x.name)
print('Completed CPU E07:',OUT,'size KB',round(OUT.stat().st_size/1024,1))
try:
 from google.colab import files
 files.download(str(OUT))
except ImportError: pass
```
