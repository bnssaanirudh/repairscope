# RSX-P2 — Independent Math Oracle and Metamorphic Benchmark

CPU-only, standalone, embedded inputs. Exhaustively verify Countdown solvability and independently recompute Arithmetic/Multistep targets for all 648 frozen problems. Produce a new, independent wording/format stress suite with ground-truth validation. **No model performance claims.**

```python
# Embedded, immutable research inputs. No upload and no old notebook rerun needed.
import base64, io, zipfile, json, hashlib
EVIDENCE_B64 = ""
EVIDENCE_SHA256 = "cd09b18ec9c37b76761e7cdfa49cca13b104d41172fe782ef168b2181dc8987d"
with zipfile.ZipFile(io.BytesIO(base64.b64decode(EVIDENCE_B64))) as _zf:
    _evidence_bytes = _zf.read("evidence.json")
assert hashlib.sha256(_evidence_bytes).hexdigest() == EVIDENCE_SHA256, 'Input corruption'
evidence = json.loads(_evidence_bytes)
problems = evidence['bank']; natural = evidence['natural']
by_id = {p['problem_id']:p for p in problems}
assert len(problems) == 648 and len(by_id)==648
assert len(natural) == 272 and len({r['trial_id'] for r in natural})==272
assert len(evidence['G_partial']) == 32 and len(evidence['H_partial']) == 12
print('Verified baseline: 648 problems; 272 historical generations; 32 G; 12 H. No generation performed.')
```

```python
import re, os, json, math, itertools, zipfile, csv, hashlib
from pathlib import Path
from fractions import Fraction
from collections import Counter
from functools import lru_cache
OUT=Path('/content/RSX_P2_ORACLE') if Path('/content').exists() else Path.cwd()/'RSX_P2_ORACLE'
OUT.mkdir(parents=True,exist_ok=True)

def solve_numeric(p):
 s=p['prompt'];t=p['template'];kind=p['kind'];nums=[int(x) for x in re.findall(r'\d+',s.split('End with')[0])]
 if kind=='arithmetic':
  assert len(nums)==2,(p['problem_id'],nums)
  return {0:lambda a,b:a+b,1:lambda a,b:a-b,2:lambda a,b:a*b,3:lambda a,b:a*b}[int(t)](*nums)
 if kind=='multistep':
  if t in (0,1):assert len(nums)==4,(p['problem_id'],nums)
  else:assert len(nums)==3,(p['problem_id'],nums)
  if t==0:return nums[0]+nums[1]*nums[2]-nums[3]
  if t==1:return nums[0]*nums[1]+nums[2]*nums[3]
  if t==2:return nums[0]*nums[1]-nums[2]
  if t==3:return nums[0]*nums[1]+nums[2]
 raise ValueError(kind)

def enumerate_countdown(values):
 """All parenthesizations/permutations with +,-,*,/ via exact rational arithmetic."""
 @lru_cache(maxsize=None)
 def recur(state):
  if len(state)==1:return {state[0]}
  out=set()
  for i in range(len(state)):
   for j in range(i+1,len(state)):
    a,b=state[i],state[j];rest=[v for k,v in enumerate(state) if k not in (i,j)]
    candidates=(a+b,a-b,b-a,a*b)
    for v in candidates:out.update(recur(tuple(sorted(rest+[v]))))
    if b:out.update(recur(tuple(sorted(rest+[a/b]))))
    if a:out.update(recur(tuple(sorted(rest+[b/a]))))
  return out
 return recur(tuple(sorted(Fraction(int(x)) for x in values)))

checks=[]
for p in problems:
 if p['kind']=='countdown':
  reachable=enumerate_countdown(p['numbers']);calc=None
  valid=(Fraction(str(p['target'])) in reachable)
  checks.append({'problem_id':p['problem_id'],'split':p['split'],'kind':p['kind'], 'template':p['template'],
       'numeric_gold':p['target'],'independent_gold':None,'gold_equal':valid,'countdown_solvable':valid,'reachable_values':len(reachable)})
 else:
  ans=solve_numeric(p)
  checks.append({'problem_id':p['problem_id'],'split':p['split'],'kind':p['kind'], 'template':p['template'],
     'numeric_gold':p['target'],'independent_gold':ans,'gold_equal':math.isclose(ans,p['target'],abs_tol=1e-8),
     'countdown_solvable':None,'reachable_values':None})
assert len(checks)==648
import pandas as pd
from IPython.display import display
checkdf=pd.DataFrame(checks);display(checkdf.groupby(['kind','split'])['gold_equal'].agg(['count','sum']))
checkdf.to_csv(OUT/'independent_math_oracle_audit.csv',index=False)
print('MISMATCHES or UNSOLVABLE:',sum(not r['gold_equal'] for r in checks))
```

```python
# Metamorphic examples are new auxiliary stress-test prompts, not part of original 648 bank.
# Only deterministic surface variations that leave every numeric quantity and operator unchanged.
NOUN_SWAP={'coins':'tokens','friend':'colleague','books':'notebooks','pens':'markers','boxes':'cartons',
 'stickers':'badges','children':'students','warehouse':'depot','packages':'parcels',
 'pencils':'crayons','flowers':'seeds','bottles':'vials','shop':'store','gardener':'farmer','cartons':'crates'}
def noun_swap(x):
 out=x
 if out.startswith('Use the numbers '):
  out=out.replace('Use the numbers ','Form an expression using the numbers ',1).replace(' exactly once to produce the target ',' each exactly once to reach the target ',1)
 for old,new in NOUN_SWAP.items():
  out=re.sub(r'\b'+re.escape(old)+r'\b',new,out,flags=re.I)
 return out

def format_swap(x):
 return x.replace('End with FINAL_ANSWER: <number>.','Finish with a separate line labeled FINAL_ANSWER: <number>.')\
         .replace('end with `FINAL_ANSWER: <number>`.','finish with a separate line labeled `FINAL_ANSWER: <number>`.')

stress=[]
for p in problems:
 for variant,fn in [('noun_swap',noun_swap),('answer_format',format_swap)]:
  altered=fn(p['prompt'])
  assert re.findall(r'-?\d+', altered)==re.findall(r'-?\d+', p['prompt']), 'Quantity accidentally changed'
  stress.append({'aux_id':p['problem_id']+'|aux|'+variant,'source_problem_id':p['problem_id'],
    'kind':p['kind'],'split':p['split'],'variant':variant,'modified':altered!=p['prompt'],
    'prompt':altered,'gold':p['target'],'exploratory_aux_only':True})
assert len(stress)==1296 and len({s['aux_id'] for s in stress})==len(stress)
with (OUT/'metamorphic_stress_prompts.jsonl').open('w') as f:
 for x in stress:f.write(json.dumps(x,sort_keys=True)+'\n')
with (OUT/'stress_suite_manifest.json').open('w') as f:json.dump({'base_problems':648,'total_aux':len(stress),
 'modified':sum(x['modified'] for x in stress),'source_sha256':EVIDENCE_SHA256,
 'purpose':'New exploratory input benchmark only; no model response has been generated for these cases.'},f,indent=2)
print('Stress prompts:',len(stress),'Modified:',sum(x['modified'] for x in stress))
```

```python
report={'independent_gold_checked':len(checks),'gold_mismatches':sum(not c['gold_equal'] for c in checks),
 'countdown_unsolvable':sum(c['kind']=='countdown' and not c['countdown_solvable'] for c in checks),
 'metamorphic_prompts':len(stress),'modified_metaprompts':sum(s['modified'] for s in stress),
 'not_model_evaluations':True}
(OUT/'summary.json').write_text(json.dumps(report,indent=2))
archive=Path('/content/RSX_P2_ORACLE_RESULTS.zip') if Path('/content').exists() else Path.cwd()/'RSX_P2_ORACLE_RESULTS.zip'
with zipfile.ZipFile(archive,'w',zipfile.ZIP_DEFLATED) as z:
 for p in sorted(OUT.rglob('*')):
  if p.is_file():z.write(p,p.relative_to(OUT))
print('RESULTS:',archive,report)
try:
 from google.colab import files;files.download(str(archive))
except ImportError:pass
```
